```python
# ============================================================
# FINSURE AI
# PHASE 3 TEST
# DATA PREPROCESSING
# ============================================================

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

from backend.data_loader import (
    DataLoader
)

from backend.preprocessor import (
    DataPreprocessor
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

DATASET_PATH = (
    PROJECT_ROOT
    / "data"
    / "loan_data.csv"
)

loader = DataLoader(
    DATASET_PATH
)

df = loader.load_data()

# ------------------------------------------------------------
# PREPROCESS
# ------------------------------------------------------------

preprocessor = (
    DataPreprocessor()
)

result = (
    preprocessor.process(
        df
    )
)

# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("=" * 60)
print("PHASE 3")
print("PREPROCESSING RESULTS")
print("=" * 60)

print()

print(
    "Training Rows:",
    result["rows_train"]
)

print(
    "Testing Rows:",
    result["rows_test"]
)

print()

print(
    "Feature Names:"
)

for feature in result[
    "feature_names"
]:

    print(
        "-",
        feature
    )

print()

print(
    "X_train Shape:",
    result["X_train"].shape
)

print(
    "X_test Shape:",
    result["X_test"].shape
)

print()

print(
    "y_train Shape:",
    result["y_train"].shape
)

print(
    "y_test Shape:",
    result["y_test"].shape
)

print()

print("=" * 60)
print("PHASE 3 COMPLETED")
print("=" * 60)
```
