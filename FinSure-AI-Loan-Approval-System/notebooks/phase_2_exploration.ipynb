```python
# ============================================================
# FINSURE AI
# PHASE 2
# DATA EXPLORATION & BUSINESS ANALYSIS
# ============================================================

from pathlib import Path
import sys

import pandas as pd

# ------------------------------------------------------------
# PROJECT PATH
# ------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

# ------------------------------------------------------------
# IMPORT
# ------------------------------------------------------------

from backend.data_loader import (
    DataLoader
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

DATASET_PATH = (
    PROJECT_ROOT
    / "data"
    / "loan_data.csv"
)

loader = DataLoader(
    DATASET_PATH
)

df = loader.load_data()

# ============================================================
# BASIC INFORMATION
# ============================================================

print("=" * 60)
print("DATASET SHAPE")
print("=" * 60)

print(df.shape)

print()

print("=" * 60)
print("COLUMN INFORMATION")
print("=" * 60)

print(df.dtypes)

# ============================================================
# TARGET ANALYSIS
# ============================================================

print()
print("=" * 60)
print("LOAN APPROVAL DISTRIBUTION")
print("=" * 60)

approval_counts = (
    df["Loan_Approved"]
    .value_counts()
)

approval_percentage = (
    df["Loan_Approved"]
    .value_counts(normalize=True)
    * 100
)

print()

print("Counts:")

print(
    approval_counts
)

print()

print("Percentages:")

print(
    approval_percentage
    .round(2)
)

# ============================================================
# NUMERICAL SUMMARY
# ============================================================

print()
print("=" * 60)
print("NUMERICAL SUMMARY")
print("=" * 60)

print(
    df.describe()
)

# ============================================================
# EMPLOYMENT ANALYSIS
# ============================================================

print()
print("=" * 60)
print("EMPLOYMENT STATUS ANALYSIS")
print("=" * 60)

employment_analysis = pd.crosstab(

    df["Employment_Status"],

    df["Loan_Approved"],

    normalize="index"

) * 100

print(
    employment_analysis.round(2)
)

# ============================================================
# APPROVED APPLICANTS
# ============================================================

approved = (
    df[
        df["Loan_Approved"] == 1
    ]
)

rejected = (
    df[
        df["Loan_Approved"] == 0
    ]
)

print()
print("=" * 60)
print("APPROVED APPLICANTS")
print("=" * 60)

print(
    f"Average Income: "
    f"{approved['Income'].mean():,.2f}"
)

print(
    f"Average Credit Score: "
    f"{approved['Credit_Score'].mean():.2f}"
)

print(
    f"Average Loan Amount: "
    f"{approved['Loan_Amount'].mean():,.2f}"
)

# ============================================================
# REJECTED APPLICANTS
# ============================================================

print()
print("=" * 60)
print("REJECTED APPLICANTS")
print("=" * 60)

print(
    f"Average Income: "
    f"{rejected['Income'].mean():,.2f}"
)

print(
    f"Average Credit Score: "
    f"{rejected['Credit_Score'].mean():.2f}"
)

print(
    f"Average Loan Amount: "
    f"{rejected['Loan_Amount'].mean():,.2f}"
)

# ============================================================
# CORRELATION
# ============================================================

print()
print("=" * 60)
print("CORRELATION WITH LOAN APPROVAL")
print("=" * 60)

numeric_df = df.select_dtypes(
    include="number"
)

correlations = (
    numeric_df.corr()[
        "Loan_Approved"
    ]
    .sort_values(
        ascending=False
    )
)

print(
    correlations
)

# ============================================================
# PHASE COMPLETE
# ============================================================

print()
print("=" * 60)
print("PHASE 2 COMPLETED")
print("=" * 60)
```
