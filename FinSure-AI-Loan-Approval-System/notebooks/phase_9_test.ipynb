```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

from backend.data_loader import DataLoader
from backend.preprocessor import DataPreprocessor
from backend.model_trainer import ModelTrainer

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

df = DataLoader(
    PROJECT_ROOT / "data" / "loan_data.csv"
).load_data()

# ------------------------------------------------------------
# PREPROCESS
# ------------------------------------------------------------

processed = DataPreprocessor().process(df)

# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

trainer = ModelTrainer()

result = trainer.train(
    processed["X_train"],
    processed["y_train"],
    processed["X_test"],
    processed["y_test"],
)

print(result)
```

```python
# ============================================================
# PHASE 9 TEST
# TRAINING PIPELINE TEST
# ============================================================

from pathlib import Path
import sys

# ------------------------------------------------------------
# PROJECT ROOT
# ------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

# ------------------------------------------------------------
# IMPORT PIPELINE
# ------------------------------------------------------------

from backend.training_pipeline import (
    TrainingPipeline
)

# ------------------------------------------------------------
# DATASET PATH
# ------------------------------------------------------------

DATASET_PATH = (
    PROJECT_ROOT
    / "data"
    / "loan_data.csv"
)

# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

pipeline = TrainingPipeline(

    dataset_path=DATASET_PATH,

    models_directory=
    PROJECT_ROOT / "models",

    artifacts_directory=
    PROJECT_ROOT / "artifacts",

)

result = pipeline.train()

# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

print("=" * 60)
print("TRAINING COMPLETED")
print("=" * 60)

print()

print(
    "Best Model:",
    result["best_model"]
)

print()

print(
    "Accuracy:",
    result["accuracy"]
)

print()

print(
    "Precision:",
    result["precision"]
)

print()

print(
    "Recall:",
    result["recall"]
)

print()

print(
    "F1 Score:",
    result["f1_score"]
)

print()

print(
    "ROC AUC:",
    result["roc_auc"]
)

print()

print(
    "Model Path:"
)

print(
    result["model_path"]
)

print()

print("=" * 60)
print("PHASE 9 COMPLETED")
print("=" * 60)
```
