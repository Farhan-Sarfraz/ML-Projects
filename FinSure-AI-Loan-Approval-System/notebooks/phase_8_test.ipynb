```python
# ============================================================
# PHASE 8 TEST
# ARTIFACT MANAGER
# ============================================================

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

from sklearn.preprocessing import (
    LabelEncoder
)

from backend.artifact_manager import (
    ArtifactManager
)

# ------------------------------------------------------------
# CREATE ENCODER
# ------------------------------------------------------------

encoder = LabelEncoder()

encoder.fit(

    [

        "Employed",

        "Self-Employed",

        "Unemployed",

    ]

)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

manager = ArtifactManager(

    PROJECT_ROOT
    / "artifacts"

)

path = manager.save_artifact(

    encoder,

    "employment_encoder.pkl",

)

print("=" * 60)
print("ARTIFACT SAVED")
print("=" * 60)
print(path)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

loaded_encoder = (

    manager.load_artifact(

        "employment_encoder.pkl"

    )

)

print()
print("=" * 60)
print("ENCODER TEST")
print("=" * 60)

print(

    loaded_encoder.transform(

        ["Employed"]

    )

)

print()
print("=" * 60)
print("PHASE 8 COMPLETED")
print("=" * 60)
```
