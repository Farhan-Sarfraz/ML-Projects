```python
# ============================================================
# FINSURE AI
# PHASE 4 TEST
# MODEL TRAINING
# ============================================================

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

from backend.data_loader import (
    DataLoader
)

from backend.preprocessor import (
    DataPreprocessor
)

from backend.model_trainer import (
    ModelTrainer
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

DATASET_PATH = (
    PROJECT_ROOT
    / "data"
    / "loan_data.csv"
)

loader = DataLoader(
    DATASET_PATH
)

df = loader.load_data()

# ------------------------------------------------------------
# PREPROCESS
# ------------------------------------------------------------

preprocessor = (
    DataPreprocessor()
)

processed = (
    preprocessor.process(
        df
    )
)

# ------------------------------------------------------------
# TRAIN MODELS
# ------------------------------------------------------------

trainer = ModelTrainer()

result = trainer.train(

    processed["X_train"],

    processed["y_train"],

    processed["X_test"],

    processed["y_test"],

)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("=" * 60)
print("MODEL COMPARISON")
print("=" * 60)

for (
    model_name,
    model_result
) in result[
    "all_models"
].items():

    metrics = (
        model_result[
            "metrics"
        ]
    )

    print()
    print(model_name)

    print(
        "Accuracy:",
        round(
            metrics[
                "accuracy"
            ],
            4
        )
    )

    print(
        "Precision:",
        round(
            metrics[
                "precision"
            ],
            4
        )
    )

    print(
        "Recall:",
        round(
            metrics[
                "recall"
            ],
            4
        )
    )

    print(
        "F1 Score:",
        round(
            metrics[
                "f1_score"
            ],
            4
        )
    )

    print(
        "ROC AUC:",
        round(
            metrics[
                "roc_auc"
            ],
            4
        )
    )

# ------------------------------------------------------------
# BEST MODEL
# ------------------------------------------------------------

best = result[
    "best_model"
]

print()
print("=" * 60)
print("BEST MODEL")
print("=" * 60)

print(
    best[
        "model_name"
    ]
)

print()

print(
    "ROC AUC:",
    round(

        best[
            "metrics"
        ][
            "roc_auc"
        ],

        4

    )
)

print()
print("=" * 60)
print("PHASE 4 COMPLETED")
print("=" * 60)
```
