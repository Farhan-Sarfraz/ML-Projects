```python
# ============================================================
# FINSURE AI
# PHASE 6 TEST
# SAVE & LOAD MODEL
# ============================================================

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.append(
    str(PROJECT_ROOT)
)

from backend.data_loader import (
    DataLoader
)

from backend.preprocessor import (
    DataPreprocessor
)

from backend.model_trainer import (
    ModelTrainer
)

from backend.model_manager import (
    ModelManager
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

DATASET_PATH = (
    PROJECT_ROOT
    / "data"
    / "loan_data.csv"
)

loader = DataLoader(
    DATASET_PATH
)

df = loader.load_data()

# ------------------------------------------------------------
# PREPROCESS
# ------------------------------------------------------------

preprocessor = (
    DataPreprocessor()
)

processed = (
    preprocessor.process(df)
)

# ------------------------------------------------------------
# TRAIN MODEL
# ------------------------------------------------------------

trainer = ModelTrainer()

training_result = trainer.train(

    processed["X_train"],

    processed["y_train"],

    processed["X_test"],

    processed["y_test"],

)

best_model = training_result[
    "best_model"
]

# ------------------------------------------------------------
# SAVE MODEL
# ------------------------------------------------------------

manager = ModelManager(

    PROJECT_ROOT
    / "models"

)

saved_path = manager.save_model(

    best_model["model"]

)

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

loaded_model = manager.load_model()

# ------------------------------------------------------------
# TEST PREDICTION
# ------------------------------------------------------------

sample = processed[
    "X_test"
][:1]

prediction = loaded_model.predict(
    sample
)

# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

print("=" * 60)
print("MODEL SAVED")
print("=" * 60)

print(saved_path)

print()

print("=" * 60)
print("MODEL LOADED")
print("=" * 60)

print(type(
    loaded_model
))

print()

print("=" * 60)
print("TEST PREDICTION")
print("=" * 60)

print(
    prediction
)

print()

print("=" * 60)
print("PHASE 6 COMPLETED")
print("=" * 60)
```
